```python
import arcpy
import os
```

```python
workspace = r'C:\Users\q_bal\Documents\Github\arcgis_restapi_notebooks\output_json_local\Timeseries_merged_json'
arcpy.env.workspace = workspace
```

```python
list_filename = os.listdir(workspace)
```

```python
list_name_fix = [fname.replace(".json.geojson","") for fname in list_filename]
```

```python
list_name_fix
```

```python
basedir_output = r"C:\Users\q_bal\Documents\Github\arcgis_restapi_notebooks\output_json_local\Timeseries.gdb"

all_path_output = [os.path.join(basedir_output,fix_name) for fix_name in list_name_fix]
```

```python
# all_path_output
```

```python
all_file = [os.path.join(workspace,path_file) for path_file in list_filename]
```

```python
for i in range(len(all_file)):
    arcpy.conversion.JSONToFeatures(
                in_json_file=all_file[i], # use the nested item take part of the group
                out_features=all_path_output[i],
                geometry_type="POLYGON" # hard coded into polygon, next time maybe need to change this if the feature is line, or point
            )
    print(f'converted to gdb feature class in {all_path_output[i]}')
```
