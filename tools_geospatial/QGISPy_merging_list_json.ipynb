```python
dir_root = r'C:\Users\q_bal\Documents\Github\arcgis_restapi_notebooks\output_json_local\Timeseries'

from os import walk

folders = []
files = []
layer = 1
w = walk(dir_root)
for (dirpath, dirnames, filenames) in w:
    #print(dirpath)
    #files = [f for f in os.listdir(dirpath) if os.path.isdir(f)]
    #print(dirnames)
    for dirname in dirnames:
        # exclude if layer more than 2, or in sub folder, we only take first layer folders
        if layer >= 2:
            break
        #print(layer)
        #print(dirname)
        new_path = os.path.join(dirpath, dirname)
        folders.append(new_path)
        files_json = [os.path.join(new_path,f) for f in os.listdir(new_path) 
                     if os.path.isfile(os.path.join(new_path,f))]
        #print(files)
        #print('\n')
        
        # let's just use it later, so that we can run in two cell
        files.append(files_json)
             
    layer += 1
```

```python
folders_error = []
files_error = []

for i in range(len(folders_name)):
    a = i+1
    folder_name = folders_name[i].replace('output_json_','')
    print(f'processing num {a} out of {len(folders_name)}')
    print(f'processing_merge of {folder_name}')
#     file_output = os.path.join(save_to,f'{folder_name}.gdb')
    file_output = os.path.join(save_to,f'{folder_name}.json')
    
    try:
        processing.run("native:mergevectorlayers", 
                       {'LAYERS':files[i],
                        'CRS':QgsCoordinateReferenceSystem('EPSG:4326'),
                        'OUTPUT':file_output})
        print(f'merged saved in {file_output}')
        print('----------------------------- \n')
    except Exception as e:
        print(f'error in {file_output} with code {e}')
        folders_error.append(folders_name[i])
        files_error.append(files[i])
        print('----------------------------- \n')
        
```
